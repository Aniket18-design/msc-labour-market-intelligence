```python
import pandas as pd

skills = pd.read_excel(
    r"D:\MScProjects\labour-market-intelligence\data\raw\onet\Essential Skills.xlsx"
)

print(skills.shape)
skills.head()
```

```python
print(skills.columns.tolist())
```

```python
skills["Title"].unique()[:20]
```

```python
skills[
    skills["Title"].str.contains(
        "Software|Cyber|Data|Database|Statistic",
        case=False,
        na=False
    )
][["O*NET-SOC Code","Title"]].drop_duplicates()
```

```python
selected_skills = skills[
    skills["Title"].isin([
        "Software Developers",
        "Information Security Analysts",
        "Data Scientists",
        "Database Administrators",
        "Statisticians"
    ])
].copy()

print(selected_skills.shape)
selected_skills.head()
```

```python
selected_skills = selected_skills[
    [
        "Title",
        "Element Name",
        "Scale Name",
        "Data Value"
    ]
]

selected_skills.head()
```

```python
selected_skills.to_csv(
    r"D:\MScProjects\labour-market-intelligence\data\processed\onet_selected_skills.csv",
    index=False
)
```
