```python
import pandas as pd

master_df = pd.read_csv(
    r"D:\MScProjects\labour-market-intelligence\data\processed\master_dataset.csv"
)

master_df["Month"] = pd.to_datetime(master_df["Month"])

master_df.head()
```

```python
dashboard_kpis = pd.DataFrame({
    "Metric": [
        "Total Job Demand",
        "Average Salary",
        "Highest Salary",
        "Highest Demand"
    ],
    "Value": [
        master_df["Demand"].sum(),
        round(master_df["Median_Salary"].mean(), 2),
        master_df["Median_Salary"].max(),
        master_df["Demand"].max()
    ]
})

dashboard_kpis
```

```python
dashboard_kpis.to_csv(
    r"D:\MScProjects\labour-market-intelligence\data\processed\dashboard_kpis.csv",
    index=False
)
```

```python
monthly_dashboard = (
    master_df
    .groupby(["Month", "Occupation"])
    .agg(
        Demand=("Demand", "sum"),
        Median_Salary=("Median_Salary", "mean")
    )
    .reset_index()
)

monthly_dashboard.head()
```

```python
monthly_dashboard.to_csv(
    r"D:\MScProjects\labour-market-intelligence\data\processed\monthly_dashboard.csv",
    index=False
)
```

```python
occupation_dashboard = (
    master_df
    .groupby("Occupation")
    .agg(
        Average_Demand=("Demand", "mean"),
        Average_Salary=("Median_Salary", "mean"),
        Maximum_Demand=("Demand", "max"),
        Maximum_Salary=("Median_Salary", "max")
    )
    .reset_index()
)

occupation_dashboard
```

```python
occupation_dashboard.to_csv(
    r"D:\MScProjects\labour-market-intelligence\data\processed\occupation_dashboard.csv",
    index=False
)
```

```python
trend_dashboard = (
    master_df
    .sort_values("Month")
)

trend_dashboard.to_csv(
    r"D:\MScProjects\labour-market-intelligence\data\processed\trend_dashboard.csv",
    index=False
)
```

```python
# ============================================================
# FINAL POWER BI PREPARATION — STEP 1
# Audit final validated datasets
# ============================================================

import pandas as pd
from pathlib import Path

project_path = Path(
    r"D:\MScProjects\labour-market-intelligence"
)

processed_path = project_path / "data" / "processed"

final_files = {
    "UK Labour Market":
        "integrated_digital_demand_salary_2023_2025.csv",

    "UK Summary":
        "final_uk_labour_market_summary.csv",

    "Regional Labour Market":
        "regional_labour_market_final_2023_2025.csv",

    "Skills + Labour Market":
        "integrated_skills_labour_market_evidence.csv",

    "O*NET Skill Profiles":
        "onet_occupation_skill_profiles.csv",

    "Technology Indicators":
        "digital_technology_demand_indicators.csv",

    "Leading Technologies":
        "leading_digital_technologies.csv",

    "Evidence Coverage":
        "final_evidence_coverage_matrix.csv"
}

print("FINAL POWER BI DATASET AUDIT")
print("=" * 80)

for name, filename in final_files.items():

    path = processed_path / filename

    print(f"\n{name}")
    print("-" * 60)
    print("File:", filename)
    print("Exists:", path.exists())

    if path.exists():
        temp = pd.read_csv(path)

        print("Shape:", temp.shape)
        print("Columns:")
        print(temp.columns.tolist())

        print("Missing values:")
        print(temp.isna().sum().sum())
```

```python
# ============================================================
# FINAL POWER BI PREPARATION — STEP 2
# Create dedicated Power BI data folder
# ============================================================

powerbi_path = processed_path / "powerbi"

powerbi_path.mkdir(parents=True, exist_ok=True)

print("Power BI folder:")
print(powerbi_path)
```

```python
# ============================================================
# FINAL POWER BI PREPARATION — STEP 3
# Export dashboard-ready datasets
# ============================================================

powerbi_files = {
    "uk_labour_market.csv":
        processed_path / "integrated_digital_demand_salary_2023_2025.csv",

    "uk_summary.csv":
        processed_path / "final_uk_labour_market_summary.csv",

    "regional_labour_market.csv":
        processed_path / "regional_labour_market_final_2023_2025.csv",

    "skills_labour_market.csv":
        processed_path / "integrated_skills_labour_market_evidence.csv",

    "onet_skill_profiles.csv":
        processed_path / "onet_occupation_skill_profiles.csv",

    "technology_indicators.csv":
        processed_path / "digital_technology_demand_indicators.csv",

    "leading_technologies.csv":
        processed_path / "leading_digital_technologies.csv",

    "evidence_coverage.csv":
        processed_path / "final_evidence_coverage_matrix.csv"
}

for output_name, source_file in powerbi_files.items():

    df = pd.read_csv(source_file)

    output_file = powerbi_path / output_name

    df.to_csv(output_file, index=False)

    print(
        f"{output_name:<30} "
        f"{df.shape[0]:>4} rows × {df.shape[1]:>2} columns"
    )

print("\nAll Power BI datasets exported successfully.")
print("Location:", powerbi_path)
```
