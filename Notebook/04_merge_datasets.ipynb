```python
import pandas as pd

demand = pd.read_csv(
    r"D:\MScProjects\labour-market-intelligence\data\processed\digital_demand_2023_2025.csv"
)

demand["Month"] = pd.to_datetime(demand["Month"])

demand.head()
```

```python
demand = demand.rename(columns={
    "2134 Programmers and software development professionals":
        "Programmers and software development professionals",

    "2135 Cyber security professionals":
        "Cyber security professionals",

    "2433 Actuaries, economists and statisticians":
        "Actuaries, economists and statisticians",

    "3133 Database administrators and web content technicians":
        "Database administrators and web content technicians",

    "3544 Data analysts":
        "Data analysts"
})
```

```python
demand_long = demand.melt(
    id_vars="Month",
    var_name="Occupation",
    value_name="Demand"
)

print(demand_long.shape)
demand_long.head()
```

```python
import pandas as pd

salary_long = pd.read_csv(
    r"D:\MScProjects\labour-market-intelligence\data\processed\salary_2023_2025.csv"
)

salary_long["Month"] = pd.to_datetime(salary_long["Month"])

print(salary_long.shape)
salary_long.head()
```

```python
print(salary_long["Occupation"].unique())
```

```python
master_df = pd.merge(
    demand_long,
    salary_long,
    on=["Month", "Occupation"],
    how="inner"
)

print(master_df.shape)
master_df.head()
```

```python
output_path = r"D:\MScProjects\labour-market-intelligence\data\processed\master_dataset.csv"

master_df.to_csv(output_path, index=False)

print("Master dataset saved successfully!")
```
