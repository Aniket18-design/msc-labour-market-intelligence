```python
import pandas as pd
import zipfile
from io import BytesIO

ons_path = Path(
    r"D:\MScProjects\labour-market-intelligence\data\raw\ons"
)

files = {
    2024: "Work Geography Table 7.7a   Annual pay - Gross 2024.xlsx",
    2025: "PROV - Work Geography Table 7.7a   Annual pay - Gross 2025.xlsx"
}

city_raw = {}

for year, filename in files.items():

    zip_path = ons_path / f"ONS_ASHE_Local_Authority {year}.zip"

    with zipfile.ZipFile(zip_path, "r") as z:
        excel_bytes = z.read(filename)

    city_raw[year] = pd.read_excel(
        BytesIO(excel_bytes),
        sheet_name="All",
        header=None
    )

    print(year, city_raw[year].shape)
```

```python
city_codes = {
    "London": "E12000007",
    "Manchester": "E08000003",
    "Birmingham": "E08000025"
}

city_tables = []

# Add the 2023 table we already created
city_tables.append(city_salary_2023)

# Extract 2024 and 2025
for year in [2024, 2025]:
    df = city_raw[year]

    selected = df[
        df[1].isin(city_codes.values())
    ][[0, 1, 3]].copy()

    selected.columns = [
        "City",
        "Area_Code",
        "Median_Annual_Salary"
    ]

    selected["Year"] = year

    city_tables.append(selected)

# Combine all three years
city_salary = pd.concat(
    city_tables,
    ignore_index=True
)

city_salary = city_salary.sort_values(
    ["City", "Year"]
).reset_index(drop=True)

city_salary
```

```python
# Make sure Year is the same datatype everywhere
city_salary["Year"] = pd.to_numeric(
    city_salary["Year"],
    errors="coerce"
).astype(int)

# Create Birmingham baseline
baseline = (
    city_salary.loc[
        city_salary["City"].str.strip() == "Birmingham",
        ["Year", "Median_Annual_Salary"]
    ]
    .copy()
    .rename(columns={
        "Median_Annual_Salary": "Birmingham_Baseline"
    })
)

print("Baseline:")
print(baseline)

# Merge
city_salary = city_salary.drop(
    columns=[
        "Birmingham_Baseline",
        "Salary_Premium_GBP",
        "Salary_Premium_Pct"
    ],
    errors="ignore"
)

city_salary = city_salary.merge(
    baseline,
    on="Year",
    how="left"
)

# Calculate premiums
city_salary["Salary_Premium_GBP"] = (
    city_salary["Median_Annual_Salary"]
    - city_salary["Birmingham_Baseline"]
)

city_salary["Salary_Premium_Pct"] = (
    city_salary["Salary_Premium_GBP"]
    / city_salary["Birmingham_Baseline"]
    * 100
).round(2)

city_salary
```

```python
from pathlib import Path

output_path = Path(
    r"D:\MScProjects\labour-market-intelligence\data\processed\city_salary_2023_2025.csv"
)

city_salary.to_csv(
    output_path,
    index=False
)

print("Saved:", output_path)
print("Shape:", city_salary.shape)
```
