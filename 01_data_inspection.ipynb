```python
import pandas as pd

file_path = r"D:\MScProjects\labour-market-intelligence\data\raw\ons\ONS_Textkernel_Job_Adverts.xlsx"

excel_file = pd.ExcelFile(file_path)

excel_file.sheet_names
```

```python
region_raw = pd.read_excel(file_path, sheet_name="1.Region NSA", header=None)

region_raw.head(15)
# pd.read_excel() → reads Excel data into a Pandas DataFrame.
# sheet_name= → chooses the worksheet.
# header=None → tells Pandas not to treat any row as column headings yet.
# region_raw → stores the result.
# Why? You are first inspecting the raw structure of the sheet.
```

```python
occupation_raw = pd.read_excel(
    file_path,
    sheet_name="3.Occupation 4-Digit NSA",
    header=None
)

occupation_raw.head(15)
```

```python
# Get the occupation names from row 4
occupation_names = occupation_raw.iloc[4].dropna().astype(str)

# Search for occupations potentially relevant to our project
keywords = [
    "data",
    "information technology",
    "IT ",
    "software",
    "computer",
    "cyber",
    "security",
    "statistic",
    "technology"
]

matches = occupation_names[
    occupation_names.str.contains(
        "|".join(keywords),
        case=False,
        regex=True
    )
]

for occupation in matches:
    print(occupation)
```

```python
# Read the sheet correctly using row 4 as the header
occupation_df = pd.read_excel(
    file_path,
    sheet_name="3.Occupation 4-Digit NSA",
    header=4
)

# Convert Month to date
occupation_df["Month"] = pd.to_datetime(occupation_df["Month"])

# Keep data from 2023 onwards
occupation_2023 = occupation_df[
    occupation_df["Month"] >= "2023-01-01"
].copy()

# Select occupations relevant to our research
selected_columns = [
    "Month",
    "2134 Programmers and software development professionals",
    "2135 Cyber security professionals",
    "2433 Actuaries, economists and statisticians",
    "3133 Database administrators and web content technicians",
    "3544 Data analysts"
]

digital_demand = occupation_2023[selected_columns]

digital_demand.head()
```

```python
# Convert occupation columns to numeric
for col in digital_demand.columns[1:]:
    digital_demand[col] = pd.to_numeric(
        digital_demand[col],
        errors="coerce"
    )

# Check data types again
digital_demand.dtypes
```

```python
# Show rows where at least one selected occupation has a missing value
missing_rows = digital_demand[
    digital_demand.iloc[:, 1:].isna().any(axis=1)
]

missing_rows
```

```python
# Re-read original data without converting occupation columns
original_df = pd.read_excel(
    file_path,
    sheet_name="3.Occupation 4-Digit NSA",
    header=4
)

original_df["Month"] = pd.to_datetime(original_df["Month"])

original_selected = original_df[
    original_df["Month"].isin(missing_rows["Month"])
][selected_columns]

original_selected
```

```python
# Main analysis period: January 2023 to May 2025
digital_demand_main = digital_demand[
    (digital_demand["Month"] >= "2023-01-01") &
    (digital_demand["Month"] <= "2025-05-01")
].copy()

print("Rows and columns:", digital_demand_main.shape)
print("Start:", digital_demand_main["Month"].min())
print("End:", digital_demand_main["Month"].max())

print("\nMissing values:")
print(digital_demand_main.isnull().sum())
```

```python
output_path = r"D:\MScProjects\labour-market-intelligence\data\processed\digital_demand_2023_2025.csv"

digital_demand_main.to_csv(output_path, index=False)

print("Saved successfully!")
print(output_path)
```
